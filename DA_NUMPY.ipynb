```python
arr = np.random.randn(5)#randn()method is used to in this case the mean is always =0
#and the standard deviation is always in =1
print(arr)
print(arr.mean())
print(arr.std())
```

```python
#np.choice()
arr = np.array([10,20,30,40,50])
result = np.random.choice(arr)
print(result)
```

```python
#choice by random
arr = np.array([10,20,30,40,50])
result = np.random.choice(arr,2)
print(result)
```

```python
#choice vth the highest probabilty
arr = [10,20,30]
result= np.random.choice(
    arr,
    size=5,
    p=[0.2,0.5,0.3]
)
print(result)
```
