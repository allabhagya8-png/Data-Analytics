```python
#multiple aggregations
import pandas as pd
employees = {
    "EmpID" : [101,102,103,104],
    "Name" : ["Asha",None,"Meena","Sai"],
    "Department" : ["Sales","It","HR",None],
    "Salary" : [45000,60000,None,89000]
}
I = [901,902,903,904]
df = pd.DataFrame(employees,index = I)
print(df)
df.isna()
df.isna().sum()
df=df.fillna({
    "Name" : "Bhagya",
    "DEpartment" : "HR",
    "Salary" : 600000
})
print(df)
df.groupby('Department')['Salary'].mean()#single aggregations
df.groupby('Department')['Salary'].max()
df.groupby('Department')['Salary'].min()
df.groupby('Department')['Salary'].count()
df.groupby('Department')['Salary'].agg(['min','max','mean'])#multiple aggregations
df.groupby('Department').agg({
    'EmpID' :['mean','max'],
    'Salary':['sum','count']
})

```

```python
import pandas as pd

data = {
   "VisitID": [501, 502, 503, 504, 505,
               506, 507, 508, 509, 510, 510, 511],
   "Patient": ["Rahul", "Anita", "Suresh", "Priya", "Kiran",
               "Anjali", "Ravi", "Meena", "Arun", "Priya",
               "Priya", "Kiran"],
   "Department": ["Cardiology", "Neurology", "Orthopedic", "General",
                  "Dental", "Cardiology", "Neurology", "General",
                  "Dental", "Cardiology", "Cardiology", "Dental"],
   "BillAmount": [5000, 7500, 4500, 2000, 1500,
                  6000, 7000, 2500, 1800, 6000, 6000, 1500]}

df = pd.DataFrame(data)
print("--- Original DataFrame ---")
print(df)

# 1. Check duplicated rows

print(df.duplicated())



# 2 Find duplicate Patient VisitIDs
duplicates = df[df["VisitID"].duplicated(keep=False)]

print(duplicates)
#3
duplicates = df[df["VisitID"].duplicated(keep="first")]
print(duplicates)
#4
duplicates = df[df["BillAmount"].duplicated(keep="last")]
total = duplicates["BillAmount"].sum()
print(duplicates)

#5
duplicates = df[df["VisitID"].duplicated(keep="last")]
print(duplicates)
#6
duplicates = df[df.duplicated()]
print(duplicates)
#7
df = df.drop_duplicates(
   subset=["VisitID"],
   keep="first")
#8
df = df.drop_duplicates(
    subset=["Patient"],
    keep="first")
#9
df = df.drop_duplicates(subset=["Patient","Department"],keep = "first")
print(df)
#10
df.drop_duplicates(["VisitID"],keep="first")
print(df)

```
